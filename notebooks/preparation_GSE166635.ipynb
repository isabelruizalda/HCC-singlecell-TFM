```python
# Imports

import os
import scanpy as sc
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import mmread
```

```python
# Definir rutas para no perderme

DATA_DIR = "../data"
RESULTS_DIR = "../results"
FIGURES_DIR = "../figures"
```

```python
dataset_path = f"{DATA_DIR}/GSE166635"

import os
os.listdir(dataset_path)
```

```python
# Cargar HCC1

X1 = mmread(os.path.join(dataset_path, "GSM5076749_HCC1.matrix.mtx")).tocsr().T

barcodes1 = pd.read_csv(
    os.path.join(dataset_path, "GSM5076749_HCC1.barcodes.tsv"),
    header=None,
    sep="\t"
)[0].tolist()

features1 = pd.read_csv(
    os.path.join(dataset_path, "GSM5076749_HCC1.features.tsv"),
    header=None,
    sep="\t"
)

adata1 = sc.AnnData(
    X=X1,
    obs=pd.DataFrame(index=barcodes1),
    var=pd.DataFrame(index=features1[1].astype(str).tolist())
)

adata1.obs["batch"] = "HCC1"
adata1.var_names_make_unique()

print(adata1)
print(adata1.shape)
```

```python
# Cargar HCC2

X2 = mmread(os.path.join(dataset_path, "GSM5076750_HCC2.matrix.mtx")).tocsr().T

barcodes2 = pd.read_csv(
    os.path.join(dataset_path, "GSM5076750_HCC2.barcodes.tsv"),
    header=None,
    sep="\t"
)[0].tolist()

features2 = pd.read_csv(
    os.path.join(dataset_path, "GSM5076750_HCC2.features.tsv"),
    header=None,
    sep="\t"
)

adata2 = sc.AnnData(
    X=X2,
    obs=pd.DataFrame(index=barcodes2),
    var=pd.DataFrame(index=features2[1].astype(str).tolist())
)

adata2.obs["batch"] = "HCC2"
adata2.var_names_make_unique()

print(adata2)
print(adata2.shape)
```

```python
# Unir muestras

adata = adata1.concatenate(adata2, join="outer")

print(adata)
print(adata.shape)
adata.obs["batch"].value_counts()
```

```python
# Guardar 
adata.write(f"{DATA_DIR}/adata_GSE166635_raw.h5ad")
```
