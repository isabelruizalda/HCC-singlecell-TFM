```python
import pandas as pd
import gseapy as gp

df = pd.read_excel("/beegfs/home/iruizdealda/HCC_singlecell_project/170_genes/170genes_clasificados_reclasificado.xlsx")

gene_col = df.columns[0]
cat_col  = df.columns[1]  # tu clasificación GO manual — se conserva
genes = df[gene_col].dropna().astype(str).tolist()

# Mapeo de grupos generales (igual que en analyze_datasets.py)
reagrupacion_general = {
    "ECM / Fibrosis / Colágeno":                 "ECM / Estroma / Remodelacion",
    "Proteolisis / remodelacion proteica":        "ECM / Estroma / Remodelacion",
    "Adhesion / ECM / migracion":                "ECM / Estroma / Remodelacion",
    "Glicosilacion / proteoglicanos / secrecion": "ECM / Estroma / Remodelacion",
    "Chemotaxis":                                "Inmunidad / Inflamacion / Chemotaxis",
    "Inmunidad / inflamacion":                   "Inmunidad / Inflamacion / Chemotaxis",
    "Complemento":                               "Inmunidad / Inflamacion / Chemotaxis",
    "Angiogénesis / Vascular":                   "Angiogenesis / Vascular / Hipoxia",
    "Hipoxia / estres / apoptosis":              "Angiogenesis / Vascular / Hipoxia",
    "Señalizacion celular":                      "Señalizacion / Regulacion celular",
    "Desarrollo / regulacion":                   "Señalizacion / Regulacion celular",
    "Transporte / canales / trafico":            "Señalizacion / Regulacion celular",
    "Neuronal / sinapsis":                       "Otros / poco clara",
    "Otros / funcion poco clara":                "Otros / poco clara",
}

df["Grupo_funcional"] = df[cat_col].map(reagrupacion_general)

# Función para buscar todos los pathways de un gen
def todos_los_pathways(gene, results_df, gene_set_name):
    sub = results_df[results_df["Gene_set"] == gene_set_name]
    mask = sub["Genes"].str.contains(rf"\b{gene}\b", regex=True, na=False)
    hits = sub[mask].sort_values("Adjusted P-value")
    if hits.empty:
        return "No encontrado"
    return " | ".join(hits["Term"].tolist())

msigdb_list   = []
kegg_list     = []
reactome_list = []
go_list       = []

for i, gene in enumerate(genes):
    print(f"  Gen {i+1}/{len(genes)}: {gene}")
    try:
        res = gp.enrichr(
            gene_list=[gene],
            gene_sets=[
                "MSigDB_Hallmark_2020",
                "KEGG_2021_Human",
                "Reactome_2022",
                "GO_Biological_Process_2023"
            ],
            organism="human",
            outdir=None,
            verbose=False
        ).results

        msigdb_list.append(  todos_los_pathways(gene, res, "MSigDB_Hallmark_2020"))
        kegg_list.append(    todos_los_pathways(gene, res, "KEGG_2021_Human"))
        reactome_list.append(todos_los_pathways(gene, res, "Reactome_2022"))
        go_list.append(      todos_los_pathways(gene, res, "GO_Biological_Process_2023"))

    except Exception as e:
        print(f"    Error en {gene}: {e}")
        msigdb_list.append("Error")
        kegg_list.append("Error")
        reactome_list.append("Error")
        go_list.append("Error")

df["MSigDB_Hallmark"]       = msigdb_list
df["KEGG_pathway"]          = kegg_list
df["Reactome"]              = reactome_list
df["GO_Biological_Process"] = go_list

output = "/beegfs/home/iruizdealda/HCC_singlecell_project/170_genes/170genes_anotacion_completa.xlsx"
df.to_excel(output, index=False)
print(f"\nGuardado en: {output}")

sin_msigdb   = (df["MSigDB_Hallmark"]       == "No encontrado").sum()
sin_kegg     = (df["KEGG_pathway"]          == "No encontrado").sum()
sin_reactome = (df["Reactome"]              == "No encontrado").sum()
sin_go       = (df["GO_Biological_Process"] == "No encontrado").sum()

print(f"\nGenes sin MSigDB:   {sin_msigdb}/170")
print(f"Genes sin KEGG:     {sin_kegg}/170")
print(f"Genes sin Reactome: {sin_reactome}/170")
print(f"Genes sin GO:       {sin_go}/170")
```
